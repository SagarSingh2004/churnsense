### Project Name
ChurnSense (AI-Powered Customer Churn Prediction & Analytics Web App)

---

### Overview

ChurnSense is an end-to-end machine learning application that helps businesses identify customers who are likely to stop using their services. The system uses customer data to predict churn probability and provides actionable insights through an interactive analytics dashboard.

---

### Business Problem

Customer churn is a major challenge for subscription-based and service-oriented businesses. Organizations often lack:

- Early warning systems for identifying at-risk customers
- Data-driven customer retention strategies
- Automated churn analysis and prediction
- Real-time decision-making tools

---

### Solution

ChurnSense addresses these challenges by providing:

- Customer churn prediction using machine learning
- Interactive dashboard with business KPIs and visualizations
- Single customer prediction interface
- Batch prediction through CSV uploads
- Analytics and model performance monitoring
- REST API support for integration with external systems

---

### Data Overview

Dataset : Blastchar Telco Customer Churn

Link : https://www.kaggle.com/datasets/blastchar/telco-customer-churn

Number of Records : 7043 customers
Number of Features : 21

Target Variable: Churn
- Yes = Customer Churn
- No = Customer will not churn

Purpose:
The dataset is used to train a machine learning model that predicts the probability of customer churn.
